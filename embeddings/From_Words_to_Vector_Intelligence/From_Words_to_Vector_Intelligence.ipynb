# Level 0 — The Anatomy of an Embedding

In machine learning, an embedding vector is far more than a collection of numbers.

Each position represents an independent latent feature learned from data.

Mathematically, a concept is represented as:

\[
\vec{v}=[f_1,f_2,\dots,f_n]
\]

Where:

- \(f_i\) is the magnitude of a latent feature.
- Each feature captures an abstract semantic property.

For this lab we'll manually define a 4-dimensional semantic space:

| Dimension | Semantic Meaning |
|-----------|------------------|
| f₁ | Medical Relevance |
| f₂ | Urgency |
| f₃ | Formality |
| f₄ | Positive Emotion |

Each value ranges from:

- **0.0 → Feature absent**
- **1.0 → Maximum intensity**

Unlike production LLMs—which automatically learn thousands of latent dimensions—we'll explicitly engineer these dimensions to understand the underlying mechanics.

---

# Level 1 — Constructing Semantic DNA

Let's create our own embedding space.

```python
import numpy as np

# Embedding dimensions:
# [Medical_Relevance, Urgency, Formality, Positive_Emotion]
# Values range from 0.0 to 1.0

vocabulary = {
    "diabetes": np.array([0.95, 0.40, 0.90, 0.05]),
    "emergency": np.array([0.85, 0.98, 0.70, 0.00]),
    "joyful": np.array([0.00, 0.10, 0.30, 0.95]),
    "headache": np.array([0.60, 0.30, 0.40, 0.10]),
}

print(f"Vector for 'diabetes': {vocabulary['diabetes']}")
```

Each coordinate now carries explicit semantic meaning.

Instead of simply storing a word, we've encoded its conceptual characteristics numerically.


# Level 2 — Dimensional Auditing

Because semantic meaning is embedded directly within vector coordinates, we can classify concepts by inspecting individual dimensions.

No cosine similarity.

No nearest-neighbor search.

No language model inference.

Just geometry.


```python
def analyze_concept(word, vector):
    print(f"\nAnalyzing concept: {word.upper()}")

    medical_score = vector[0]
    urgency_score = vector[1]

    if medical_score > 0.8:
        print("→ Category: Clinical / Medical")

    elif medical_score > 0.4:
        print("→ Category: General Health")

    else:
        print("→ Category: Non-Medical")

    if urgency_score > 0.8:
        print("⚠ ACTION REQUIRED: High urgency detected!")
```

```python
analyze_concept("diabetes", vocabulary["diabetes"])
analyze_concept("emergency", vocabulary["emergency"])
analyze_concept("joyful", vocabulary["joyful"])
```

This demonstrates that a semantic classifier can operate solely on latent vector coordinates.


# Level 3 — Semantic Context Manipulation

One of the most fascinating properties of embedding spaces is that semantic context can be modified mathematically.

Instead of changing the text itself, we'll adjust only specific latent dimensions.

```python
# Original concept
original_headache = vocabulary["headache"].copy()

print(f"\nOriginal vector: {original_headache}")

modified_headache = original_headache.copy()

# Increase urgency
modified_headache[1] = 0.99

# Increase formality
modified_headache[2] = 0.95

print(f"Modified vector: {modified_headache}")

print("\nSemantic transformation completed.")
```

Although the underlying concept remains *headache*, its semantic interpretation has shifted toward:

> **Critical Neurological Emergency**

This transformation occurred by modifying only two latent dimensions—without changing the original word.
